# Dropping Columns Quick Notes

1. **Drop columns / rows**

   * `df.drop(...)` → Removes specified rows or columns.
   * `axis=1` → columns | `axis=0` → rows
   * `data.drop('col1', axis=1)` → Removes a **single column**.
   * `data.drop(['col1', 'col2'], axis=1)` → Removes **multiple columns**.
   * `data.drop('Unnamed: 0', axis=1)` → Removes an unwanted index column such as `0, 1, 2, 3...` accidentally saved in a CSV.

2. **Missing values**

   * `df.dropna()` → Removes rows containing **missing (`NaN`) values**.

3. **Duplicate rows**

   * `df.drop_duplicates()` → Removes **duplicate rows**.

4. **Returned DataFrame & `inplace`**

   * `df.drop(...)`, `df.dropna()`, `df.drop_duplicates()` → By default, **return a new DataFrame**; the original DataFrame is unchanged.
   * Assign back to save the change: `df = df.dropna()`
   * `inplace=True` → Modifies the **original DataFrame directly**.
   * Examples: `df.dropna(inplace=True)`, `df.drop_duplicates(inplace=True)`, `df.drop('col1', axis=1, inplace=True)`

   **Pandas `inplace=True`**

   Simple rule:

   If the Pandas method **supports** **`inplace=True`**, then:

   ```python
   df.method(inplace=True)
   ```

   modifies the original `df`.

   But **don't assume every Pandas method supports it**.

5. **Check remaining data**

   * `len(df)` → Returns the **number of rows** remaining in the DataFrame.

## 5. How does the data look mathematically?

````python
# describes() - applies on numeric data only
df.describe()
````

#### What to look for in describe()

- `count` → Compare with total rows → identify missing values.
- `Mean` → "Where is the center?"
- `mean vs 50% (median)` → Large difference may indicate skewness.
- `min/max` → Look for extreme or unusual values / possible outliers.
- `std` → Understand How spread out are the values around that center. That number came from looking at every value's distance from the mean
- `25% and 75%` → Understand where the middle 50% of data lies
- `min → 25% → 50% → 75% → max` → Get a quick idea of how the values are distributed

### Titanic insights

- `Age`: 714 vs 891 → 177 missing values
- `Fare`: mean (32.20) >> median (14.45) → likely right-skewed because some very high fares are pulling the mean upward.
- `Age`: mean (29.7) ≈ median (28) → no strong skew indicated.
- `Age`: min (0.42) to max (80) → recorded ages range from an infant to an 80-year-old passenger.
- `Fare`: 0 → 512.33 → wide range; investigate extreme values. Here fare seems fine
- min/max alone cannot prove something is an outlier. To actually determine outliers, we'd use something like the IQR method or a box plot later
- `Age`: mean = 29.7, std = 14.5 → average age is 29.7 years, with considerable variation in ages around this average.
- `Fare`: mean = 32.20, std = 49.69, while the central 50% lies only between 7.91–31 → fares are highly spread out, with some very high fares (max = 512.33) pulling the mean upward.
- `Age`: 25% = 20 and 75% = 38 → central 50% of recorded passenger ages lies between 20 and 38 years.
- `Age`: min → 25% → 50% → 75% → max = `0.42 → 20 → 28 → 38 → 80` → gives a quick picture of the age range, center, and where the central portion of the data is concentrated.
